## 1. Motivación: ¿por qué importa el espacio?

Casi todas las células de un organismo comparten esencialmente el mismo genoma. Hay excepciones: las células sexuales tienen un único juego de cromosomas y pasan por recombinación; los linfocitos B y T reordenan segmentos de su ADN; los glóbulos rojos maduros no tienen núcleo ni ADN; y las células tumorales acumulan mutaciones.

Entonces, ¿por qué una neurona y un hepatocito (célula hepática) hacen cosas tan distintas? Una analogía útil: **es el mismo código fuente, pero con distinto estado de ejecución.** La diferencia no está en qué genes existen, sino en **cuáles se activan (se expresan), en qué medida y en qué momento**.

Ese estado depende de señales químicas que la célula recibe, y muchas provienen de su entorno inmediato: moléculas secretadas por células vecinas, contacto directo, gradientes de oxígeno o de nutrientes. Por eso, la identidad y la función de una célula están **influenciadas** por su contexto espacial: quiénes son sus vecinas, a qué distancia están y qué moléculas intercambian en un punto del espacio 2D o 3D. (También influyen el linaje y el estado intrínseco de la célula; el espacio es una pieza, no la única.)

### ¿Qué medimos?
Los métodos que veremos miden principalmente **ARN mensajero (transcritos)**, y en algunos casos proteínas, **junto con las coordenadas** de dónde se midieron. Así se habla de transcriptómica espacial y proteómica espacial. (La genómica espacial también existe, pero mide ADN, no expresión.)

## 2. La abstracción: del tejido a las estructuras de datos

Un corte de tejido puede verse como una fuente **no estructurada**. La transcriptómica espacial lo abstrae en tres componentes:

| Componente | Símbolo | Forma | Qué representa |
|---|---|---|---|
| Matriz de expresión | $\mathbf{X}$ | $N \times M$ | Abundancia de cada gen en cada unidad de muestreo |
| Coordenadas espaciales | $\mathbf{S}$ | $N \times 2$ | Posición $(x_i, y_i)$ de cada unidad |
| Imagen histológica | $\mathbf{I}$ | $H \times W \times C$ | Contexto morfológico (tinción H&E o inmunofluorescencia) |

A partir de $\mathbf{S}$ se construye además un **grafo espacial de vecindad** $G = (V, E)$, donde los nodos son las unidades de muestreo y las aristas representan proximidad física.

Usaremos **un solo dataset** durante toda la guía para ver cada componente en un objeto real.

* **Valores:** conteos enteros de moléculas antes de normalizar; valores reales después.

## 5. Componente 2: las coordenadas $\mathbf{S}$

$$\mathbf{S} = \begin{bmatrix} x_1 & y_1 \\ x_2 & y_2 \\ \vdots & \vdots \\ x_N & y_N \end{bmatrix}$$

### ¿Píxeles o micrómetros?
Las coordenadas suelen venir en **píxeles de la imagen original**. Para hablar de distancias biológicas hay que convertirlas usando un factor de escala. En Visium conocemos dos medidas físicas: el spot mide ~55 µm de diámetro y los centros de spots vecinos están a 100 µm. Cada una permite estimar cuántos µm vale un píxel. **Las dos estimaciones deberían coincidir**; la celda siguiente las compara.

````python
lib_id = list(adata.uns["spatial"].keys())[0]
sf = adata.uns["spatial"][lib_id]["scalefactors"]
print(sf)

from scipy.spatial import cKDTree

# Distancia mediana al vecino más cercano, en píxeles
d, _ = cKDTree(S).query(S, k=2)
nn_px = np.median(d[:, 1])

# Estimación A: el spot de Visium mide ~55 µm de diámetro
um_per_px_A = 55 / sf["spot_diameter_fullres"]
# Estimación B: la distancia entre centros de spots vecinos es 100 µm
um_per_px_B = 100 / nn_px

print(f"Distancia mediana al vecino más cercano: {nn_px:.0f} px")
print(f"A) por diámetro del spot (55 µm): {um_per_px_A:.3f} µm/px "
      f"→ vecino más cercano ≈ {nn_px * um_per_px_A:.0f} µm")
print(f"B) por distancia entre spots (100 µm): {um_per_px_B:.3f} µm/px "
      f"→ vecino más cercano = 100 µm por construcción")
print(f"Diferencia relativa entre A y B: {abs(um_per_px_A - um_per_px_B) / um_per_px_B:.0%}")
````

> **Cómo leer este resultado.** Si A y B difieren de forma apreciable (en una ejecución previa de este notebook, alrededor del 16 %), al menos una suposición no se cumple para *este* objeto: que las coordenadas estén en los píxeles que describe el factor de escala, o que `spot_diameter_fullres` corresponda realmente a 55 µm. Aquí no lo resolvemos. La lección es **verificar la escala contra una cantidad física conocida antes de interpretar distancias**.

## 6. Componente 3: la imagen histológica $\mathbf{I}$

Para usarla junto con $\mathbf{S}$ se necesita un **registro de coordenadas**: alinear el sistema de coordenadas de los spots con el de la imagen, típicamente mediante un factor de escala (y, en algunas tecnologías, transformaciones adicionales).

## 7. El grafo espacial de vecindad $G = (V, E)$

El grafo se construye **solo a partir de $\mathbf{S}$**: los **nodos** $V$ son las unidades de muestreo y las **aristas** $E$ representan proximidad física. La expresión $\mathbf{X}$ no participa en decidir quién es vecino de quién; se **adjunta a los nodos como atributos** (cada nodo carga su vector de expresión). Así, $\mathbf{S}$ define la *estructura* y $\mathbf{X}$ el *contenido*.

**Ojo:** el grafo *no viene dado por los datos*. Es una **decisión de modelado**. Las reglas más comunes son:

| Regla | Idea | Cuándo suele usarse |
|---|---|---|
| Rejilla (*grid*) | vecinos inmediatos en la retícula del instrumento | Visium (rejilla hexagonal, 6 vecinos) |
| k vecinos más cercanos | conecta cada nodo con sus $k$ más cercanos | datos con posiciones irregulares (células) |
| Radio fijo | conecta todo lo que esté a menos de una distancia | cuando la escala biológica importa |
| Triangulación de Delaunay | conecta según una teselación del plano | posiciones irregulares, sin elegir $k$ |

> **No confundir dos grafos sobre los mismos nodos.** El dataset ya trae `adata.obsp['connectivities']`, que se calculó a partir de la **expresión** (vecinos en el espacio del PCA): dos spots son vecinos si *se parecen* transcripcionalmente, estén donde estén. El grafo espacial de esta sección conecta spots que *están cerca en el tejido*, sin importar si se parecen. Comparar ambos es la base de muchos análisis posteriores.


````python
sq.gr.spatial_neighbors(adata, coord_type="grid", n_neighs=6)  # Visium: rejilla hexagonal
A = adata.obsp["spatial_connectivities"]
print(type(A))
print("Forma de A:", A.shape, "| entradas no nulas (cada arista cuenta 2 veces):", A.nnz)
````

````python
# Grado de cada nodo (número de vecinos)
deg = np.asarray(A.sum(axis=1)).ravel()
pd.Series(deg).value_counts().sort_index()
````

````python
# Dibujamos el grafo: completo y con zoom
from matplotlib.collections import LineCollection
from scipy.spatial import cKDTree

rows, cols = A.nonzero()
mask = rows < cols  # cada arista una sola vez (A es simétrica en la rejilla)
segs = np.stack([S[rows[mask]], S[cols[mask]]], axis=1)

# Ventana de zoom: ~12 spots de ancho, centrada en el centro del tejido
paso = np.median(cKDTree(S).query(S, k=2)[0][:, 1])
cx, cy = np.median(S, axis=0)
h = 6 * paso

fig, axes = plt.subplots(1, 2, figsize=(11, 6))
for ax, titulo, tam in zip(axes, ["Grafo completo", "Zoom: se ven las aristas"], [6, 25]):
    ax.add_collection(LineCollection(segs, linewidths=0.6, colors="gray"))
    ax.scatter(S[:, 0], S[:, 1], s=tam, c="black", zorder=3)
    ax.set_aspect("equal")
    ax.set_xlabel("x (px)")
    ax.set_ylabel("y (px)")
    ax.set_title(titulo)
axes[0].autoscale()
axes[0].invert_yaxis()  # convención de imagen: el origen está arriba a la izquierda
axes[1].set_xlim(cx - h, cx + h)
axes[1].set_ylim(cy + h, cy - h)
plt.tight_layout()
plt.show()
````

````python
# Otras reglas para el mismo tejido (guardadas con otro nombre para no sobrescribir)
sq.gr.spatial_neighbors(adata, coord_type="generic", n_neighs=6, key_added="knn6")
sq.gr.spatial_neighbors(adata, coord_type="generic", delaunay=True, key_added="delaunay")


def resumen_grafo(A):
    """Devuelve (entradas no nulas sin diagonal, aristas únicas no dirigidas)."""
    A = sp.csr_matrix(A)
    dirigidas = A.nnz - np.count_nonzero(A.diagonal())
    unicas = sp.triu(A.maximum(A.T), k=1).nnz  # i–j cuenta una vez si i→j o j→i
    return dirigidas, unicas


for k in ["spatial", "knn6", "delaunay"]:
    dirigidas, unicas = resumen_grafo(adata.obsp[f"{k}_connectivities"])
    print(f"{k:>9} → entradas no nulas: {dirigidas:>6} | aristas únicas (no dirigidas): {unicas:>6}")
````

> **¿Por qué dos columnas?** El grafo `knn6` es **dirigido**: cada nodo elige a sus 6 vecinos más cercanos, pero que *i* esté entre los 6 más cercanos de *j* no implica que *j* esté entre los de *i*. Por eso (en la ejecución previa) tuvo exactamente $N \times k = 2688 \times 6 = 16{,}128$ entradas no nulas. Las rejillas y Delaunay son **simétricos**: cada arista aparece dos veces en la matriz. Para comparar reglas de forma justa hay que contar **aristas únicas no dirigidas**.

## 8. El contenedor unificado: `AnnData` / `SpatialExperiment`

Ecosistemas como Python (`Scanpy` / `Squidpy`) y R (`Bioconductor`) guardan todo lo anterior en **un único objeto**. Así se corresponde cada componente con un atributo del `AnnData`:

| Atributo | Contiene | Componente |
|---|---|---|
| `.X` | matriz de expresión ($N \times M$) | $\mathbf{X}$ |
| `.obs` | metadatos de observaciones (una fila por spot/célula) | |
| `.var` | metadatos de genes (una fila por gen) | |
| `.obsm['spatial']` | coordenadas $(x, y)$ ($N \times 2$) | $\mathbf{S}$ |
| `.uns['spatial']` | imagen H&E y factores de escala | $\mathbf{I}$ |
| `.obsp['spatial_connectivities']` | grafo de vecindad **espacial** ($N \times N$, disperso) | $G$ (derivado de $\mathbf{S}$) |
| `.obsp['connectivities']` | grafo de vecindad en el **espacio de expresión** | derivado de $\mathbf{X}$ |
| `.obsm['X_pca']`, `.obsm['X_umap']` | proyecciones de baja dimensión de la expresión | derivados de $\mathbf{X}$ |
| `.layers` | versiones alternativas de $\mathbf{X}$ (p. ej. conteos crudos); en este objeto está vacío | |

> **Clasifica lo que ves.** No todo lo que hay en un `AnnData` es un dato original del experimento. En este objeto, **X**, **S** e **I** son las mediciones; el grafo espacial, el PCA, el UMAP y los clusters (`leiden`, `cluster`) son resultados de *análisis* sobre ellas. Distinguir dato medido de dato derivado es un hábito clave para no tratar un resultado como si fuera una medición.

**Ejercicio 6 (conceptual).** Quieres estudiar la señalización por **contacto directo** entre células vecinas. Elige una regla de vecindad (rejilla, kNN, radio fijo o Delaunay) y justifícala en cada caso: (a) un corte de Visium; (b) un corte de MERFISH con células de densidad muy variable.

<details>
<summary>Ver respuesta</summary>

**(a) Visium:** la *rejilla*. Las posiciones son regulares y el propio instrumento define quiénes son vecinos inmediatos (6 en la retícula hexagonal), así que no hay un parámetro que inventar.

**(b) MERFISH:** las posiciones son irregulares, así que la rejilla no aplica. No hay una única respuesta correcta: lo importante es que la regla elegida **coincida con la pregunta biológica** (aquí, contacto implica escala pequeña y local, lo que favorece el radio) y que conozcas su costo.
</details>


## 11. Glosario

| Término | Significado |
|---|---|
| Transcrito | molécula de ARN producida a partir de un gen |
| Expresión génica | medida de cuánto se "activa" un gen |
| Conteo | número de moléculas detectadas de un gen en una observación |
| Dispersión (*sparsity*) | proporción de ceros en la matriz |
| Spot | zona circular del chip donde se captura ARN (puede contener varias células) |
| Panel dirigido | lista acotada de genes que se miden a propósito |
| Registro de coordenadas | alineación entre el sistema de coordenadas de los spots y el de la imagen |
| `AnnData` | objeto de Python que agrupa matriz, metadatos, coordenadas e imagen |
| Grafo dirigido / no dirigido | en un grafo dirigido la arista $i \to j$ no implica $j \to i$ (p. ej. kNN); en uno no dirigido sí |
| Grafo espacial vs. de expresión | el primero conecta spots **cercanos en el tejido** (se construye de $\mathbf{S}$); el segundo conecta spots **parecidos en expresión** (se construye de $\mathbf{X}$) |
| Datos crudos vs. normalizados | conteos enteros tal como salen del instrumento vs. valores ya escalados o transformados (p. ej. log1p) |
| Dato medido vs. derivado | medido: $\mathbf{X}$, $\mathbf{S}$, $\mathbf{I}$; derivado: resultado de un análisis sobre ellos (PCA, UMAP, clusters, grafos) |
